# 🚀 Python Avançado com Visualizações

Este notebook foca em gráficos, visualizações e exemplos mais avançados.

## 📊 1. Gráfico de Linha

```python
import matplotlib.pyplot as plt

x = list(range(10))
y = [i**2 for i in x]

plt.plot(x, y)
plt.title("Crescimento Quadrático")
plt.xlabel("X")
plt.ylabel("X²")
plt.show()
```

## 📈 2. Gráfico de Dispersão

```python
import random
x = [random.randint(0, 50) for _ in range(50)]
y = [random.randint(0, 50) for _ in range(50)]

plt.scatter(x, y)
plt.title("Dispersão Aleatória")
plt.show()
```

## 📉 3. Histograma

```python
dados = [random.gauss(0, 1) for _ in range(1000)]

plt.hist(dados, bins=30)
plt.title("Distribuição Normal")
plt.show()
```

## 🧮 4. NumPy - Computação Rápida

```python
import numpy as np

matriz = np.random.rand(3,3)
print("Matriz:")
print(matriz)

print("Transposta:")
print(matriz.T)
```

## 📊 5. Pandas - Análise de Dados

```python
import pandas as pd

dados = {
    "Nome": ["Ana", "João", "Maria"],
    "Idade": [23, 35, 29]
}

df = pd.DataFrame(dados)
print(df)

print("\nMédia de idade:", df["Idade"].mean())
```

## 🤖 7. Machine Learning Visual

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

X, _ = make_blobs(n_samples=200, centers=3)

kmeans = KMeans(n_clusters=3)
kmeans.fit(X)

plt.scatter(X[:,0], X[:,1])
plt.title("Clusters (KMeans)")
plt.show()
```

## 🎨 8. Gráfico Animado

```python
import matplotlib.animation as animation

fig, ax = plt.subplots()
x = []
y = []

def update(frame):
    x.append(frame)
    y.append(frame**2)
    ax.clear()
    ax.plot(x, y)

ani = animation.FuncAnimation(fig, update, frames=range(10))
plt.show()
```
